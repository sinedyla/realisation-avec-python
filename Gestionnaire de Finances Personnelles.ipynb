```python
import numpy as np
import sympy as sp
import math

def calculatrice_financiere():
    P = float(input("Dépôt périodique (P) : "))
    r = float(input("Taux d'intérêt annuel (%) : ")) / 100
    t = float(input("Nombre d'années (t) : "))
    choix_croissance = input("Croissance (1: Mensuelle, 2: Continue) : ")

    if choix_croissance == '1':
        annuite = P * (((1 + r/12)**(12*t) - 1) / (r/12))
    else:
        annuite = (P / r) * (math.exp(r * t) - 1)
    print(f"Valeur de l'annuité : {annuite:,.2f}")

    P_m = float(input("\nMontant du prêt (Hypothèque) : "))
    r_m = float(input("Taux annuel (%) : ")) / 100
    t_m = int(input("Durée (années) : "))
    i = r_m / 12
    n = t_m * 12
    mensualite = P_m * (i * (1 + i)**n) / ((1 + i)**n - 1)
    print(f"Paiement mensuel : {mensualite:,.2f}")

    initial = float(input("\nÉpargne actuelle : "))
    depot = float(input("Dépôt mensuel : "))
    r_ret = float(input("Taux attendu (%) : ")) / 100
    t_ret = float(input("Années avant la retraite : "))
    solde = initial * (1 + r_ret/12)**(12*t_ret) + depot * (((1 + r_ret/12)**(12*t_ret) - 1) / (r_ret/12))
    print(f"Estimation solde retraite : {solde:,.2f}")

    r_doub = float(input("\nTaux pour doublement (%) : ")) / 100
    temps = math.log(2) / math.log(1 + r_doub)
    print(f"Temps nécessaire pour doubler : {temps:.2f} ans")

    print("\nRésolution log(x, base) = b")
    base = float(input("Base du log : "))
    b_val = float(input("Valeur de b : "))
    x_sym = sp.symbols('x')
    sol_log = sp.solve(sp.Eq(sp.log(x_sym, base), b_val), x_sym)
    print(f"Solution x : {sol_log}")

    num = float(input("\nNombre à convertir en notation scientifique : "))
    print(f"Résultat : {num:.2e}")

calculatrice_financiere()
```
