```python
import matplotlib.pyplot as plt
import random

def jeu_points():
    score = 0
    difficultes = {"1": 5, "2": 10, "3": 20}
    choix = input("Difficulté (1: Facile, 2: Moyen, 3: Difficile): ")
    limite = difficultes.get(choix, 5)

    for _ in range(3):
        x_cible = random.randint(-limite, limite)
        y_cible = random.randint(-limite, limite)

        plt.figure(figsize=(5,5))
        plt.plot(x_cible, y_cible, 'ro')
        plt.grid(True)
        plt.axhline(0, color='black', lw=1)
        plt.axvline(0, color='black', lw=1)
        plt.xlim(-limite-1, limite+1)
        plt.ylim(-limite-1, limite+1)
        plt.title("Quelles sont les coordonnées (x, y) ?")
        plt.show()

        try:
            ans_x = int(input("x = "))
            ans_y = int(input("y = "))
            if ans_x == x_cible and ans_y == y_cible:
                print("Correct !")
                score += 1
            else:
                print(f"Raté ! C'était ({x_cible}, {y_cible})")
        except ValueError:
            print("Entrée invalide.")

    print(f"Score final : {score}/3")

jeu_points()
```

```python
import random
import sympy as sp

def jeu_algebre():
    x = sp.symbols('x')
    diff = input("Difficulté (1: Simple, 2: Avancé): ")
    taille = 10 if diff == "1" else 50

    for i in range(3):
        a = random.randint(1, taille) * random.choice([-1, 1])
        b = random.randint(1, taille) * random.choice([-1, 1])
        c = random.randint(1, taille) * random.choice([-1, 1])

        if diff == "1":
            equation = sp.Eq(x + a, b) # x + a = b
        else:
            equation = sp.Eq(a*x + b, c) # ax + b = c

        print(f"\nProblème {i+1}: Résous {sp.latex(equation)}")
        solution = sp.solve(equation, x)[0]

        try:
            reponse = float(input("x = "))
            if reponse == float(solution):
                print("Bien joué !")
            else:
                print(f"Faux. La réponse était {solution}")
        except ValueError:
            print("Entrée non numérique.")

jeu_algebre()
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interactive
import random

# Paramètres du mur aléatoires
mur_x = random.randint(3, 8)
mur_h = random.randint(2, 7)

def trajectoire(a, b, c):
    x = np.linspace(0, 10, 100)
    y = a*x**2 + b*x + c

    plt.figure(figsize=(10, 5))
    plt.plot(x, y, label="Trajectoire")
    plt.bar(mur_x, mur_h, width=0.3, color='red', label="Mur") # Le mur

    # Vérification collision
    y_au_mur = a*mur_x**2 + b*mur_x + c
    if y_au_mur > mur_h and a < 0:
        plt.title("BRAVO ! Le projectile est passé !")
    else:
        plt.title("ÉCHEC : Ajustez la parabole")

    plt.ylim(0, 10)
    plt.xlim(0, 10)
    plt.axhline(0, color='black')
    plt.legend()
    plt.show()

# Niveau 1 : Sliders
print(f"Cible : Passer au dessus d'un mur de hauteur {mur_h} situé à x = {mur_x}")
interactive_plot = interactive(trajectoire, a=(-2.0, 0.0, 0.05), b=(0.0, 10.0, 0.1), c=(0.0, 10.0, 0.1))
display(interactive_plot)
```
