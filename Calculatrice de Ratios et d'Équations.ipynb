```python
import sympy as sp
from sympy import symbols, Eq, solve, sqrt, Integer

def calculatrice_avancee():
    x = symbols('x')
    print("--- Calculatrice Multifonction ---")
    print("1. Proportions (n1/d1 = x/d2)")
    print("2. Résoudre pour x")
    print("3. Factoriser/Simplifier une racine carrée")
    print("4. Conversions (Décimal, Fraction, Pourcentage)")

    choix = input("\nOption : ")

    if choix == '1':
        n1 = sp.Rational(input("Numérateur 1 : "))
        d1 = sp.Rational(input("Dénominateur 1 : "))
        d2 = sp.Rational(input("Dénominateur 2 : "))
        res = solve(Eq(n1/d1, x/d2), x)
        print(f"Résultat : x = {res[0]}")

    elif choix == '2':
        eq_input = input("Entrez l'équation égale à 0 (ex: 2*x - 4) : ")
        equation = sp.sympify(eq_input)
        res = solve(equation, x)
        print(f"Solutions : {res}")

    elif choix == '3':
        n = int(input("Nombre sous la racine : "))
        print(f"Forme simplifiée : {sqrt(n)}")

    elif choix == '4':
        val = input("Entrez la valeur (ex: 0.75, 3/4, ou 75%) : ")
        if '%' in val:
            num = sp.Rational(val.replace('%', '')) / 100
        else:
            num = sp.Rational(val)

        print(f"Décimal : {float(num)}")
        print(f"Fraction : {num}")
        print(f"Pourcentage : {float(num)*100}%")

calculatrice_avancee()
```
